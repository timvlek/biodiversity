<h1 style="color: #57A6DC; font-size: 2em;">Overview</h1>

<h2 style="font-size: 1.5em; font-weight: bold;">Apply hypothesis tests</h2>

If your exploration results in statistical hypotheses, consider applying a hypothesis test to determine if there is enough evidence in the data to reject the null hypothesis or not. For example, in our sample project we found that the percent of native protected species was higher than non-native, and tested for an association. Whether or not the results are significant, the outcome of a hypothesis test may provide key insights into the data questions at hand.

<details>
<summary style="color: #57A6DC; font-weight: bold;">
Hint
</summary>
________________________________________________________________________________________

In general, here are some things to keep in mind while performing hypothesis
tests:

+ Use one or two sample t-tests for numerical data.
+ Use a Chi-squared test for categorical data associations.
+ Understand the assumptions for each hypothesis test.
+ Before running the hypothesis test, make sure to clearly define your null and alternative hypothese, choose significance levels, and check your sample size.
+ Be sure to report the outcome of the hypothesis test (ex. test statistic, p-value, etc.) and determine if the results are significant or not.

Useful cheatsheets:

+ [Hypothesis testing: testing a sample statistic](https://www.codecademy.com/learn/paths/master-statistics-with-python/tracks/stats-hypothesis-testing/modules/hypothesis-testing-testing-a-sample-statistic/cheatsheet)

**Duplicate values**

15 Out of 23.296 data points in `observation.csv` are duplicates. It is difficult to identify, whether these are valid data points or duplicates. There are 21.068 combinations of <span class='variable'>park</span> and <span class='variable'>scientific name</span> with a single data point, 1.060 with two data points and 36 with three data points. This suggest the values are not duplicates, but rather different measurements. Although this raises the question why there would be multiple measurements for some combinations of <span class='variable'>park</span> and <span class='variable'>species</span> and not others? Are these errors? It appears it would be a very costly error to accidently sample the same <span class='variable'>species</span> in the same <span class='variable'>park</span> mulitple times - especially performing this error 1.060 + 36 * 2 times -. Although these measurements might have been purposefull it would be incorrect to compare 1 measurement of a species with multiple measurements for another species. Therefore, the multiple measurement will grouped and replaced with their respective averages.

```python
# Replace multiple measurements with a single mean
o_df = o_df\
    .groupby(by=['Scientific name', 'Park'])\
    .mean()\
    .reset_index(level=1)\
    .reset_index(level=0)

o_df['No. observations'] = o_df['No. observations'].astype('int64')
```

**Missing Values**

The <span class='variable'>conservation status</span> of most <span class='variable'>species</span> in `species.csv` is not provided. The 'missing' data is likely, simply because the species are not listed. The underlying reasons for the absence of a listing can be very different. A species can be widely regarded to be of no concern and therefore never have been researched as part of a census. Or they have been part of a census, but never been determined to be of any concern. Although, this is an immensily important distinction, the reason for the missing data is considered irrelevant and all missing data are labelled as <span class='variable'>not listed</span> for further analysis.

```python
# Replace missing values with 'Not listed' for further analysis.
s_df.fillna('Not listed', inplace=True)
```

All value names will be capitalized -where appropriate- for the purpose of clarity and visualization. Futhermore the designation 'National Park' is considered redundant and therefore dropped from the <span class='variable'>park</span> column for brevity.

For the purpose of visualization they are assumed to be ordered in the order of increasing endangerment as follows: <span class='variable'>species of concern</span>, <span class='variable'>threatened</span>, and <span class='variable'>endangered</span>. Unfortunately a clear definition of <span class='variable'>in recovery</span> could not be quickly identified. Here, it is considered to mean a species is <span class='variable'>threatened</span> or <span class='variable'>endangered</span>, but the <span class='variable'>conservation status</span> has recently been improving. As such, it is onsidered an all together different classification, not on the spectrum of the other three categories.

### General observations

```python
### Calculate the number of species listed and not-listed in each class

# Define rename function to relabel the conservation status not 'Not listed' to 'Listed'
rename = lambda x: 'Listed' if x != 'Not listed' else 'Not listed'

# Rename the conservation status
l_df = s_df.copy()
l_df['Conservation status'] = l_df['Conservation status'].apply(rename)

# Calculate the order of class counts
class_order = l_df['Class'].value_counts().index

# Create a crosstable to create a count of species class and conservation status in each class
ct = pd.crosstab(
    index=l_df['Class'], 
    columns=l_df['Conservation status']
)

# Sort the dataframe
t_df = ct[listed_order].reindex(class_order[::-1])

# Display the result
t_df.reindex(class_order).reset_index().style.hide(axis='index')
```

```python
# Perform a chi-square hypthesis test on all known species.
_, p, _, _ = chi2_contingency(ct)
p
```

Based on a chi-square hypothesis test, the conclusion is that the distribution
of <span style='color: #FAB733'>listed</span> vs <span style='color:
#808080'>not listed</span> species is not identical across classes.

```python
### Plot the number of species listed and not-listed in each class

# Define plot customization
listed_palette = {
    'Not listed': '#808080',
    'Listed': '#FAB733'
    }

color = listed_palette
legend = 'Conservation status:'
description = 'Stacked barplot of counts of class species vs. listed status'
title = "Most species are vascular plants and most species are not listed"
subtitle = """
Most species within each class are not listed. 
Only very few species are actually listed(fortunately).
Epecially among both vascular and non-vascular plants, most species are not listed.
"""

# Create the plot
plot = t_df.plot(
    kind='barh', 
    stacked=True, 
    edgecolor='white', 
    figsize=(8, len(t_df)*4/6),
    legend=None, 
    xlabel='', 
    ylabel='', 
    color=color
    )

# Define plot customization continued
xticks = plt.xticks()[0][:-1]
xlabels = [f'{x/1000:.0f}k' for x in xticks]

# Customize the plot
plt.grid(axis='y')
plt.legend(ncol=4, bbox_to_anchor=(.45, 1))
text = plt.text(s=title, x=0, y=9, size=14, fontweight='bold')
text = plt.annotate(subtitle, xycoords=text, xy=(0, -4))
text = plt.annotate('Conservation status:', xycoords=text, xy=(0, -.1))
plt.xticks(ticks=xticks, labels=xlabels)

# Save the plot
plt.savefig(f'./plots/{description}.svg')

# Show the plot
plt.show();
```

**Are certain types of <span class='variable'>species</span> more likely to be <span class='variable' style='color: #FAB733'>listed</span>?**

It is assumed, that the type of <span class='variable'>species</span> refers to the <span class='variable'>class</span> of <span class='variable'>species</span>, otherwise this question would be non-sensical: any individual <span class='variable'>species</span> can have only one <span class='variable'>conservation status</span>. 
Then, certain types of <span class='variable'>species</span> are more likely to be <span class='variable' style='color: #FAB733'>listed</span>. Therefore, it is highly likely, that certain types of <span class='variable'>species</span> are more likely to be <span class='variable' style='color: #FAB733'>listed</span> than others.
